````python
import requests
import pandas as pd

API_KEY = "091855a6e7msh416e5bf4bb76483p151d0ejsne6e93a69daf6"

url = "https://job-salary-data.p.rapidapi.com/salary"

headers = {
    "x-rapidapi-key": API_KEY,
    "x-rapidapi-host": "job-salary-data.p.rapidapi.com"
}

job_titles = [
    "nodejs developer",
    "software engineer",
    "frontend developer",
    "backend developer",
    "full stack developer",
    "python developer",
    "java developer",
    "react developer",
    "mobile app developer",
    "ios developer",
    "android developer",
    "data scientist",
    "data engineer",
    "data analyst",
    "machine learning engineer",
    "ai engineer",
    "business intelligence analyst",
    "devops engineer",
    "cloud architect",
    "site reliability engineer",
    "cybersecurity analyst",
    "systems administrator",
    "network engineer",
    "qa engineer",
    "product manager",
    "project manager",
    "scrum master",
    "ui ux designer",
    "product designer",
    "solutions architect",
]

all_jobs = []

for job in job_titles:

    querystring = {
        "job_title": job,
        "location": "India"
    }

    response = requests.get(
        url,
        headers=headers,
        params=querystring
    )

    print(job, response.status_code)

    if response.status_code == 200:
        data = response.json()

        if isinstance(data, list):
            all_jobs.extend(data)

        elif isinstance(data, dict):
            all_jobs.append(data)

    else:
        print("Error:", response.text)



job_df = pd.json_normalize(all_jobs)

print("\nTotal records:", len(job_df))

display(job_df)
````
